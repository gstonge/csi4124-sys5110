## Waiting times

A great deal of this course comes down to *how long until the next thing happens*.
The simplest answer — events arriving at a constant average rate, independently of
each other — gives waiting times that follow an exponential distribution.

Sample 10,000 of them at rate $\lambda = 2$ per minute and check the mean against
the theoretical $1/\lambda$.

```python
rate = 2.0                     # events per minute
waits = rng.exponential(scale=1 / rate, size=10_000)

print(f"sampled mean : {waits.mean():.4f} min")
print(f"theory 1/rate: {1 / rate:.4f} min")
```

```python
fig, ax = plt.subplots(figsize=(6, 3.5))
ax.hist(waits, bins=50, density=True, alpha=0.6, label="sampled")

t = np.linspace(0, waits.max(), 300)
ax.plot(t, rate * np.exp(-rate * t), lw=2, label=r"$\lambda e^{-\lambda t}$")

ax.set_xlabel("waiting time (min)")
ax.set_ylabel("density")
ax.legend()
fig.tight_layout()
```
