# Part 1 — The logistic equation

A population $P(t)$ that grows at rate $r$ while it is small and saturates at a carrying
capacity $K$:

$$\frac{dP}{dt} = r P \left(1 - \frac{P}{K}\right).$$

This one is worth starting with because it can be solved in closed form. From an initial
value $P_0$ at time $0$,

$$P(t) = \frac{K P_0 e^{rt}}{K + P_0\left(e^{rt} - 1\right)} .$$

Nothing singles out $t = 0$ in that formula, so the same expression maps the state at any
time $t$ to the state one time step $h$ later:

$$P(t + h) = \frac{K\, P(t)\, e^{rh}}{K + P(t)\left(e^{rh} - 1\right)} .$$

That is an **exact transition function**. It is the same object Euler's method is trying to
approximate, only without any error.

# Part 2 — A pendulum

A rigid pendulum of length $\ell$, at an angle $\theta$ from the downward vertical, obeys

$$\frac{d^2\theta}{dt^2} = -\frac{g}{\ell}\sin\theta .$$

This is a second-order ODE. Introduce the angular velocity
$\omega = d\theta/dt$ as a state variable of its own, and it becomes two coupled
first-order equations on the state $\boldsymbol{x} = [\theta, \omega]^\top$:

$$\frac{d\theta}{dt} = \omega, \qquad
  \frac{d\omega}{dt} = -\nu^2 \sin\theta, \qquad
  \nu^2 \equiv \frac{g}{\ell} .$$

$g$ and $\ell$ only ever appear through their ratio, so the model has a **single
parameter**, $\nu$ — the frequency of the small oscillations. Everything below takes
$\nu = 1$.
